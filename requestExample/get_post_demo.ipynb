```python
import requests
```

```python
# GET
```

```python
get_url = 'https://jsonplaceholder.typicode.com/todos/1'
```

```python
get_response = requests.get(get_url)
```

```python
print(get_response.json())
```

```python
#POST
```

```python
to_do_item = {'userId' : 2, 'title' : 'my to do', 'completed' : False}
post_url = 'https://jsonplaceholder.typicode.com/todos'
post_response = requests.post(post_url, json=to_do_item)
```

```python
print(post_response.json())
```

```python
#Optinonal header
headers = {'Content-Type': 'application/json'}
post_response = requests.post(post_url, json=to_do_item, headers=headers)
```

```python
print(post_response.json())
```

```python
import json
#We manually convert the data to JSON format and send it using data.
```

```python
post_response = requests.post(post_url, data=json.dumps(to_do_item), headers=headers)
```

```python
print(post_response.json())
```
