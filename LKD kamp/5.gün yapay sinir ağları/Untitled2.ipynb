```python
import numpy as np
import matplotlib.pyplot as plt
from minisom import MiniSom
from sklearn.datasets import make_blobs
```

```python
data, y = make_blobs(n_samples=1000, centers=4, cluster_std=2, random_state=31)
plt.scatter(data[:,0], data[:,1], c="black", s=100, alpha=1, label="Original Data")
```

```python
# 10x10 SOM oluştur
som = MiniSom(x=10, y=10, input_len=2, sigma=1.0, learning_rate=0.7)
som.random_weights_init(data)
som.train_random(data, 000) # 5000 itirasyon
```

```python
plt.figure(figsize=(6,6))
plt.scatter(data[:,0], data[:,1], c="lightblue", label="Veri")
for i in range(10):
    for j in range(10):
        w = som.get_weights()[i,j]
        plt.plot(w[0], w[1], "ro")

plt.title("Kohonen Ağı/SOM Ağı Haritası")
plt.legend()
plt.show()
```
