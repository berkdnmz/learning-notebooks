```python
import numpy as np
import random
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1],
])
```

```python
y = np.array([0,1,1,1])
```

```python
np.random.seed(31)
weights = np.random.randn(2)
bias = 0
lr = 0.1
epochs = 100
```

```python
def step_function(x):
    return 1 if x >= 0 else 0
```

```python
for epoch in range(epochs):
    for xi, target in zip(X,y):
        net = np.dot(xi, weights) + bias
        y_pred = step_function(net)

        error = target - y_pred

        weights += lr * error * xi
        bias += lr * error
```

```python
print("Eğitim tamamlandı")
```
