```python
import numpy as np
import random
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x * (1 - x) # buradaki c sigmoidin çıktısı
```

```python
X = np.array([[0,0],
             [0,1],
             [1,0],
             [1,1]])

y = np.array([[0], [1], [1], [0]])
```

```python
#rastgele ağırlıklar
np.random.seed(31)
input_neurons = 2
hidden_neurons = 2
output_neurons = 1

w1 = np.random.randn(input_neurons, hidden_neurons) # giriş-> gizli/ara
b1 = np.zeros((1, hidden_neurons))

w2 = np.random.randn(hidden_neurons, output_neurons) # gizli -> çıktı
b2 = np.zeros((1, output_neurons))

lr = 0.1
```

```python
epochs = 5000
for epoch in range(epochs +1):
    net1 = np.dot(X, w1) + b1
    fnet1 = sigmoid(net1)

    net2 = np.dot(fnet1, w2) + b2
    fnet2 = sigmoid(net2)

    # loss (MSE)
    loss = np.mean((y - fnet2) ** 2)

    # backpropagation
    # çıktı katmanı hatası
    deltaw2 = (y - fnet2) * sigmoid_derivative(fnet2)

    # ara katman
    deltaw1 = np.dot(deltaw2, w2.T) * sigmoid_derivative(fnet1)

    w2 += lr * np.dot(fnet1.T, deltaw2)
    b2 += lr * np.sum(deltaw2, axis=0, keepdims=True)

    w1 += lr * np.dot(X.T, deltaw1)
    b1 += lr * np.sum(deltaw1, axis=0, keepdims=True)

    if (epoch + 1) % 1000 == 0:
        print(f"Epoch: {epoch}, Loss: {loss: .4f}")

print(f"\nTahminler: ")
print(fnet2)
```
